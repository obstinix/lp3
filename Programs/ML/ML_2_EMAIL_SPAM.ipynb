```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.metrics import classification_report, accuracy_score, confusion_matrix, precision_score, recall_score, f1_score

df = pd.read_csv("emails.csv")
df.head()
```

```python
X = df.drop(columns=['Email No.', 'Prediction'])
X.head()
```

```python
Y = df['Prediction']
```

```python
Xtrain, Xtest, Ytrain, Ytest = train_test_split(X, Y, test_size=0.2, random_state=42)
```

```python
# Training the KNN model to classify the spam emails
knn_model = KNeighborsClassifier(n_neighbors=5)
knn_model.fit(Xtrain, Ytrain)
```

```python
knn_prediction = knn_model.predict(Xtest)
knn_accuracy = accuracy_score(Ytest, knn_prediction)
knn_accuracy
```

```python
knn_cm = confusion_matrix(Ytest, knn_prediction)
knn_cm
```

```python
knn_cr = classification_report(Ytest, knn_prediction)
print(knn_cr)
```

```python
# Training the SVM model for spam classification
svm_model = SVC(kernel="linear")
svm_model.fit(Xtrain, Ytrain)
```

```python
svm_prediction = svm_model.predict(Xtest)
svm_accuracy = accuracy_score(Ytest, svm_prediction)
svm_accuracy
```

```python
svm_cm = confusion_matrix(Ytest, svm_prediction)
svm_cm
```

```python
svm_cr = classification_report(Ytest, svm_prediction)
print(svm_cr)
```

```python
# Creating a Comparison Table for the two models to analyse which model performed the best
results = {
    'Model': ['KNN', 'SVM'],
    'Accuracy': [
        knn_accuracy,
        svm_accuracy
    ],
    'Precision': [
        precision_score(Ytest, knn_prediction),
        precision_score(Ytest, svm_prediction)
    ],
    'Recall': [
        recall_score(Ytest, knn_prediction),
        recall_score(Ytest, svm_prediction)
    ],
    'F1 Score': [
        f1_score(Ytest, knn_prediction),
        f1_score(Ytest, svm_prediction)
    ]
}
comparison = pd.DataFrame(results)
print(comparison)

# By analysing the comparison results of the two models we can infer that
# SVM is the better model than KNN.
```
